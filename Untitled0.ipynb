```python
# [ 예측 전용 코드 ]
import pandas as pd
from statsmodels.tsa.api import SimpleExpSmoothing

try:
    # 1. 데이터 준비
    file_path = '/content/GS25 SAMPLE.csv'
    df = pd.read_csv(file_path)

    # 2. 컬럼 이름 및 데이터 타입 설정 (이전과 동일)
    actual_date_col = '날짜'    #  실제 날짜 컬럼 이름
    actual_demand_col = '판매량' # 실제 수요 컬럼 이름

    df[actual_date_col] = pd.to_datetime(df[actual_date_col])
    df = df.set_index(actual_date_col)
    demand_series = df[actual_demand_col]

    # 3. 모델 학습 (파일에 있는 전체 90일 데이터 사용)
    ses_model = SimpleExpSmoothing(demand_series, initialization_method="estimated").fit()

    # 4. 미래 7일 예측
    future_forecast = ses_model.forecast(7)

    print("--- 학습 완료 후, 미래 7일 수요 예측 결과 ---")
    print(future_forecast.round(2))

except FileNotFoundError:
    print(f"오류: '{file_path}' 파일을 찾을 수 없습니다.")
except KeyError as e:
    print(f"오류: 컬럼 이름을 찾을 수 없습니다 - {e}. 코드의 컬럼 이름을 확인해주세요.")
except Exception as e:
    print(f"알 수 없는 오류가 발생했습니다: {e}")
```
